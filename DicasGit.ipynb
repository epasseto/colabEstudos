### Fluxo de trabalho do Git

- criando o diretório dos repositórios

```python
!git init
```

- por padrão, o Git **não vê** os nossos arquivos!

  - este é o momento em que a gente está **escrevendo código**

  - para isso, use `stage`

```python
!git add index.html
```

- `commit` é a versão que guardamos do nosso projeto

  - mexe no código → dá `stage` em tudo → dá `commit` para salvar a versão

    - `commit1`: `git add index.html`

    - `commit2`: mexe no código + `git add index.html`

    - `commit3`: " + `git add index.html`

    - ⋯



- VsCode já é integrado com Git

  - aba `source control`

  - `A` significa adicionado (`U` não adicionado)

```python
!git commit -m "inicial"
!clear
!git commit -m "adicionando um cabeçalho" #mensagem de identificação do commit
```

- `style.css`

### Traçando uma linha do tempo

    commit inicial ⇒ estilo cSS ⇒ ...

Para voltar na linha do tempo, eu preciso pegar o `hash` do `commit` que eu quero retornar:

```python
!git switch --detach ...
```

- se eu der um `log`, aparece apenas o primeiro:

  - notação `master` ←a `branch` que está sendo evoluída (este é o nome padrão, se houver apenas uma)

  - notação `HEAD` ←o que está atualmente vigente

  - notação `HEAD -> master`

```python
!git log
```

- para mostrar todos:

```python
!git log --all
```

- esse dá um `ID` reduzido (mais fácil de usar):

```python
!git log --oneline --all
```

```python
!git switch master
```

### Trabalhando com `branches`

- para exibir todas as nossas `branches`

- nossa `branch` ativa fica marcada

```python
!git branch
```

Fizemos o `commit` no vermelho

- exibe o grafo das `branches`

```python
!git log --online --all --graph
```

`Merge`

- aplicar a alteração de uma `branch` naquela principal

- misturamos duas branches

```python
!git switch master
```

```python
!git merge vermelho
```

Conflito:

- tanto a branch principal como a vermelho alteram uma mesma característica (a cor do parágrafo)

*Nem sempre a operação de merge envolve resolução de conflitos!*

```python
!git add .
```

```python
!git commit -m "merge com vermelho"
```

```python
!git log --oneline --all --graph
```

Agora apagamos o ramo que foi mergeado

```python
!git branch -D vermelho
```

Trabalho colaborativo:

```python
!git branch nova-feature
```

Modifico...

```python
!git add .
```

```python
!git commit -m "Parágrafos em azul"
```

Push, agora falando qual a `branch` a ser carregada:

- `origin` ←repositório remoto

- `nova-feature` ←repositório local, com alterações

Main remoto, agora foi atualizado com a branch nova-feature

```python
!git push origin nova-feature
```

Por princípio, em um repositório remoto compartilhado, eu JAMAIS altero a branch principal!



Agora em `Pull-requests`, seleciono `main` e `nova-feature`:

- eu preciso fazer um `merge`

- o Tech Leader verifica e aprova/rejeita

- quem faz o Pull-request é quem fez a alteração

- eu solicito uma revisão de código, a ser aplicada ao `main`

- o dono do repositório recebe a notificação, avalia em `files-changed` e clica em `Merge pull request`, e `Confirm merge`

ANTES de dar um `push`, sempre dê um `pull` antes!

- ele serve para atualizar pequenas alterações realizadas por terceiros no projeto, antes de carregar o novo pacote

- normalmente isso não vai requerer revisão manual, por as pessoas costumam trabalhar paralelamente em módulos diferentes de um mesmo sistema


```python
!git switch main
```

Agora eu preciso atualizar as outras branches, além da principal:

```python
!git pull
```

```python
!git log --online --all
```
